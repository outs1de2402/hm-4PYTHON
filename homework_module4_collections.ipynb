## Калькулятор статистики
Дано список оцінок:
```python
grades = [85, 92, 78, 90, 88, 76, 95, 89, 84, 91]
```

Обчисліть:
1. Середнє значення
2. Медіану (середній елемент відсортованого списку)
3. Мінімум та максимум
4. Діапазон (різниця між max та min)
5. Скільки оцінок вище середнього

````python
grades = [85, 92, 78, 90, 88, 76, 95, 89, 84, 91]
sum = 0
for i in grades:
    sum += i
a = sorted(grades)
ser = sum / len(grades)
print(ser)
n = len(a)
if  n% 2 == 1:
    m = a[n // 2]
else:
    m = (a[n // 2 - 1] + a[n // 2]) / 2
print(m)
mi = min(grades)
ma = max(grades)
print(mi, ma)
print(ma - mi)
suma = []
for i in grades:
    if i > ser:
        suma.append(i)
print(len(suma))
````
